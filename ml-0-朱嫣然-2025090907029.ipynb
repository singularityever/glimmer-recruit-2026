```python
import numpy as np #支持大量的维度数组与矩阵运算

A=np.random.randint(1, 101, size=(10, 10)) #10*10 1<=x<101
print("矩阵A:\n", A)
print("-"*50)

B=np.eye(10) #生成10*10的单位矩阵

C=np.dot(A, B) #A*B
print("矩阵C:\n", C)
print("-"*50)

A[A>80]=0 #筛选替换

A1 = A[0:5, 0:5]    # 左上角 5x5
A2 = A[0:5, 5:10]   # 右上角 5x5
A3 = A[5:10, 0:5]   # 左下角 5x5
A4 = A[5:10, 5:10]  # 右下角 5x5, 均是小取等大不取

# 使用 .mean() 求均值， 没写参数，默认计算整个数组平均
print("左上角 5x5 矩阵的均值:", A1.mean())
print("右上角 5x5 矩阵的均值:", A2.mean())
print("左下角 5x5 矩阵的均值:", A3.mean())
print("右下角 5x5 矩阵的均值:", A4.mean())
```

```python
a = torch.tensor([[1], [2], [3]])  # shape (3, 1)
b = torch.tensor([4, 5, 6])        # shape (3,)
# 计算 c = a + b，并写出 c 的形状
c = a+b  #形状：（3，3）
#b可以补为（1，3），从右往左看，其中一个为1，广播为3
print("矩阵 c 的值:\n", c)
print("c 的形状:", c.shape)
print("-" * 30)
# 计算 d = a * b，并解释广播规则 -->逐元素乘法
d = a*b
print("矩阵 d 的值:\n", d)
#广播规则：张量形状不一致时，从右往左对齐，缺一维就补一维
#1、从最右边的维度开始对齐。
#2、依次比较每个维度，两个维度相等，或者其中一个为 1，那么这一维可以广播。
#3、如果其中一个维度缺失，就把它当成 1，再继续对齐。
```

```python
import torch
a = torch.tensor([[1, 2],
                  [3, 4]])
b = torch.tensor([[5, 6]])
# 将 a 和 b 按行拼接（什么样的可以按列拼接？如何按列拼接？或者还有其他维度的拼接）
c = torch.cat((a, b),dim=0) #第0维度即为行方向；行数相同可以列拼接（dim=1）；可以在其他维度上堆叠张量，拼成立体张量
print(c)
# 补全代码并且写出c的形状
#形状：（3，2）
print("c 的形状:", c.shape)
```
